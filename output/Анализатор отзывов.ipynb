# Домашнее задание: анализатор отзывов

Решение семи частей задания. Все функции работают с отзывами вида `(текст, оценка)`.

## Исходные данные

Список сохраняет порядок отзывов, а каждый кортеж хранит неизменяемую пару текста и оценки.

```python
reviews = [
    ("Отличный товар быстрая доставка", 5),
    ("Товар хороший но доставка медленная", 4),
    ("Ужасный товар не советую никому", 1),
    ("Доставка быстрая товар нормальный", 4),
    ("Отличный магазин отличный сервис", 5),
    ("Плохой товар сломался сразу", 2),
]
```

## Часть 1. Нормализация текста

```python
def normalize(text):
    if not isinstance(text, str):
        return ""
    return text.lower().strip().replace(",", " ").replace(".", " ")
```

## Часть 2. Частота слов

```python
def word_frequency(reviews):
    counter = {}
    for text, rating in reviews:
        for word in normalize(text).split():
            counter[word] = counter.get(word, 0) + 1
    return counter
```

## Часть 3. Уникальные слова

```python
def unique_words(reviews):
    words = set()
    for text, rating in reviews:
        for word in normalize(text).split():
            words.add(word)
    return words
```

## Часть 4. Частые слова

```python
def top_words(reviews, n):
    frequencies = word_frequency(reviews)
    sorted_words = sorted(frequencies.items(), key=lambda item: item[1], reverse=True)
    return sorted_words[:max(0, n)]
```

## Часть 5. Статистика оценок

```python
def rating_stats(reviews):
    ratings = [rating for text, rating in reviews]
    stats = {
        "average": 0,
        "max": 0,
        "min": 0,
        "positive": 0,
        "negative": 0,
        "neutral": 0,
    }
    if not ratings:
        return stats

    stats["average"] = sum(ratings) / len(ratings)
    stats["max"] = max(ratings)
    stats["min"] = min(ratings)
    for rating in ratings:
        if rating >= 4:
            stats["positive"] += 1
        elif rating <= 2:
            stats["negative"] += 1
        else:
            stats["neutral"] += 1
    return stats
```

## Часть 6. Текстовый отчёт

Числа в примере отчёта в условии расходятся с исходными отзывами. Здесь все показатели вычисляются по данным.

```python
def make_report(reviews):
    words = unique_words(reviews)
    stats = rating_stats(reviews)
    lines = [
        "ОТЧЕТ ПО ОТЗЫВАМ",
        "",
        f"Всего отзывов: {len(reviews)}",
        f"Уникальных слов: {len(words)}",
        f"Средняя оценка: {stats['average']:.2f}",
        "",
        f"Положительных: {stats['positive']}",
        f"Нейтральных: {stats['neutral']}",
        f"Отрицательных: {stats['negative']}",
        "",
        "Топ-3 слова:",
    ]
    for word, count in top_words(reviews, 3):
        lines.append(f"  {word} — {count}")
    return "\n".join(lines)


print(make_report(reviews))
```

## Проверка функций

Проверки охватывают примеры из условия, пустые данные и запрос топа больше числа слов.

```python
assert normalize("Отличный, товар.") == "отличный  товар "
assert normalize("  БЫСТРАЯ доставка  ") == "быстрая доставка"
assert normalize("Товар, хороший. Очень!") == "товар  хороший  очень!"
assert normalize(None) == ""

sample = [("Товар хороший", 5), ("Товар плохой", 1)]
assert word_frequency(sample) == {"товар": 2, "хороший": 1, "плохой": 1}
assert unique_words([("товар хороший товар", 5), ("товар плохой", 1)]) == {"товар", "хороший", "плохой"}
assert top_words([("товар товар товар", 5), ("доставка доставка", 4), ("сервис", 5)], 2) == [("товар", 3), ("доставка", 2)]
assert len(top_words(sample, 10)) == 3

assert rating_stats([("текст", 5), ("текст", 4), ("текст", 1), ("текст", 3)]) == {
    "average": 3.25, "max": 5, "min": 1,
    "positive": 2, "negative": 1, "neutral": 1,
}
assert rating_stats([]) == {
    "average": 0, "max": 0, "min": 0,
    "positive": 0, "negative": 0, "neutral": 0,
}
assert word_frequency([]) == {}
assert unique_words([]) == set()
assert top_words([], 3) == []
assert "Средняя оценка: 0.00" in make_report([])
print("Все проверки пройдены")
```

## Часть 7. Интерактивное меню

Запустите последнюю ячейку для работы с меню. Для завершения введите `0`.

```python
def run():
    while True:
        print("\n1 — Полный отчет")
        print("2 — Топ слов")
        print("3 — Статистика оценок")
        print("0 — Выход")
        choice = input("Выберите действие: ").strip()

        if choice == "1":
            print(make_report(reviews))
        elif choice == "2":
            for word, count in top_words(reviews, 3):
                print(f"  {word} — {count}")
        elif choice == "3":
            stats = rating_stats(reviews)
            print(f"Средняя оценка: {stats['average']:.2f}")
            print(f"Максимальная оценка: {stats['max']}")
            print(f"Минимальная оценка: {stats['min']}")
            print(f"Положительных: {stats['positive']}")
            print(f"Нейтральных: {stats['neutral']}")
            print(f"Отрицательных: {stats['negative']}")
        elif choice == "0":
            print("До свидания!")
            break
        else:
            print("Неверный выбор. Введите 1, 2, 3 или 0.")


run()
```
